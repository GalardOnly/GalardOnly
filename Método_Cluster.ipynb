```python
from sklearn import datasets
from sklearn.metrics import confusion_matrix
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
```

```python
iris = datasets.load_iris()
iris
```

```python
def Plot_clusters(data, labels, title):
    colors = ['red','green','purple','black']
    plt.figure(figsize=(8,4))
    for i,c,l in zip(range(-1,3), colors, ['Noise', 'Setosa', 'Versicola','Viriginica']):
        if i == -1:
            plt.scatter(data[labels == i, 0], data[labels == i, 3], c=colors[i], label = 1, alpha =0.5, s=50, marker='x')
        else:
            plt.scatter(data[labels == i, 0], data[labels == i, 3], c=colors[i], label = 1, alpha =0.5, s=50, marker='o')
        plt.legend()
        plt.title(title)
        plt.xlabel('comprimento da sépala')
        plt.ylabel('largura da pétala')
        plt.show()
```

```python
iris.target
```

```python
kmeans = KMeans(n_clusters=3,n_init='auto')
kmeans.fit(iris.data)
print(kmeans.labels_)
resultados = confusion_matrix(iris.target, kmeans.labels_)
print(resultados)


Plot_clusters(iris.data,kmeans.labels_,'Cluster KMeans' )
```

```python
dbscan = DBSCAN(eps=0.5,min_samples=3)
dbscan_labels = dbscan.fit_predict(iris.data)
print(dbscan_labels)


Plot_clusters(iris.data,dbscan_labels, 'Cluster DBScan')
```

```python
agglo = AgglomerativeClustering(n_clusters=3)
aggloe_labels = agglo.fit_predict(iris.data)
print(aggloe_labels)
resultados = confusion_matrix(iris.target, aggloe_labels)
print(resultados)

Plot_clusters(iris.data,aggloe_labels, 'Cluster Hierarquico')
```

```python
plt.figure(figsize=(12,6))
plt.title("cluster hierarquico: dendograma")
plt.xlabel("indice")
plt.ylabel("distancia")
linkage_matrix = linkage(iris.data, method='ward')
dendrogram(linkage_matrix, truncate_mode='lastp',p=15)
plt.axhline(y=7, c='gray', lw=1, linestyle='dashed')
plt.show()
```
