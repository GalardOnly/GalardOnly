```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
from sklearn.feature_selection import chi2, SelectKBest
```

```python
anuncio = pd.read_csv('ad.data', header=None)#se o arquivo nao tiver cabeçalho usamos o header
anuncio.shape
```

```python
#iremos separar a variavel dependente da variavel independente
x = anuncio.iloc[:,:-1].values #aqui eu puxei todas as linhas, ':-1' menos a ultima
y = anuncio.iloc[:,-1].values # são todas as linhas com a ultima
```

```python
x
```

```python
y #o Y é a classe, adnon a pessoa nao clicou no anuncio
```

```python
x_treinamento, x_teste, y_treinamento, y_teste = train_test_split(x,y, test_size=0.3, random_state=0)
```

```python
modelo1 = GaussianNB()
modelo1.fit(x_treinamento, y_treinamento)
previsoes1 = modelo1.predict(x_teste)
print(accuracy_score(y_teste, previsoes1))
```

```python
#agora faremos a seleção de atributos para criar um novo modelo

selecao = SelectKBest(chi2, k=7)
x_novo = selecao.fit_transform(x,y)
```

```python
x_novo
```

```python
x_novo.shape
```

```python
print(selecao.get_support())
```

```python
x_treinamento, x_teste, y_treinamento, y_teste = train_test_split(x_novo,y, test_size=0.3, random_state=0)
```

```python
modelo2 = GaussianNB()
modelo2.fit(x_treinamento, y_treinamento)
previsoes2 = modelo2.predict(x_teste)
print(accuracy_score(y_teste, previsoes2))

#apos fazer a seleção com os subconjuntos ganhos quase 30 de performance
```
