```python
from sklearn import datasets
import numpy as np
from sklearn.preprocessing import scale
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
from hopkins import * #usar o * significa que teremos todas as funções disponiveis
from metric import *
from visual_assessment_of_tendency import *
```

```python
import warnings #aqui desativamos possiveis avisos do codigo para nao deixar ele poluido
warnings.filterwarnings('ignore', category=FutureWarning)
```

```python
iris = datasets.load_iris()
clust1 = scale(iris.data)
clust2 = scale(np.random.rand(150,4))#estamos gerando uma matriz com 150 linhas e 4 colunas aleatorio
```

```python
print(clust2)
```

```python
#construindo o grafico elbow

inertia = []
for i in range(1,8):
  kmeans = KMeans(n_clusters=i, n_init='auto')
  kmeans.fit(clust1)
  inertia.append(kmeans.inertia_)
plt.plot(range(1,8), inertia)
plt.title("elbow")
plt.show()
```

```python
#construindo o grafico elbow

inertia = []
for i in range(1,8):
  kmeans = KMeans(n_clusters=i, n_init='auto')
  kmeans.fit(clust2)
  inertia.append(kmeans.inertia_)
plt.plot(range(1,8), inertia)
plt.title("elbow")
plt.show()
```
